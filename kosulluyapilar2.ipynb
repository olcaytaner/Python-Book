```
kilo = int(input("Kac kilosun?"))
boy = float(input("Boyun kac?"))
vucutKitleEndeksi = kilo / (boy * boy)
if vucutKitleEndeksi > 30:
	print("Cok sismansin")
elif vucutKitleEndeksi > 25:
	print("Sismansin")
elif vucutKitleEndeksi > 20:
	print("Normalsin")
else:
	print("Zayifsin")
```

```
avro = int(input("Euro degerini girin:"))
paraBirimi = input("Dolar mi lira mi:")
if paraBirimi == "dolar":
	dolar = avro * 1.12
	print("Dolar: " + str(dolar))
elif paraBirimi == "lira":
	lira = avro * 3.24
	print("Lira: " + str(lira))
else:
	print("Dolar veya lira girmeliydiniz!!!")
```

```
hangiDereceden = input("Cevireceginiz Sicaklik Derece Birimini Girin C, K veya F:")
hangiDereceye = input("Cevrilecek Sicaklik Derece Birimini Girin C, K veya F:")
derece = int(input("Dereceyi Girin:"))
if hangiDereceden == "F" and hangiDereceye == "C":
	C = 5 * (derece - 32) / 9
	print("C:" + str(C))
if hangiDereceden == "F" and hangiDereceye == "K":
	K = 5 * (derece - 32) / 9 + 273
	print("K:" + str(K))
if hangiDereceden == "C" and hangiDereceye == "F":
	F = 1.8 * derece + 32
	print("F:" + str(F))
if hangiDereceden == "C" and hangiDereceye == "K":
	K = derece + 273
	print("K:" + str(K))
if hangiDereceden == "K" and hangiDereceye == "C":
	C = derece - 273
	print("C:" + str(C))
if hangiDereceden == "K" and hangiDereceye == "F":
	F = 1.8 * (derece - 273) + 32
	print("F:" + str(F))
```
