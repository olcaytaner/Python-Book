```
kilo = int(input("Kac kilosun?"))
boy = float(input("Boyun kac?"))
vucutKitleEndeksi = kilo / (boy * boy)
print(vucutKitleEndeksi)
if vucutKitleEndeksi > 30:
	print("Cok kilolusun")
if 25 < vucutKitleEndeksi < 30:
	print("Hafif kilolusun")
if 20 < vucutKitleEndeksi < 25:
	print("Normalsin")
if vucutKitleEndeksi < 20:
	print("Hafif zayifsin")
```
