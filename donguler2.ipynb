```
toplam = 0
ogrenciSayisi = int(input("Ogrenci sayisini girin:"))
for i in range(ogrenciSayisi):
	print(str(i + 1) + ". ogrencinin notunu girin:")
	ogrenciNot = int(input())
	toplam = toplam + ogrenciNot
if ogrenciSayisi > 0:
	print("Ogrencilerin ortalamasi:" + str(toplam / ogrenciSayisi))
else:
	print("Ortalamayi hesaplayacak kadar ogrenci yok")
```

```
toplam = 0
i = 1
ogrenciNot = int(input("1. ogrencinin notunu girin:"))
while ogrenciNot > 0:
	toplam = toplam + ogrenciNot
	i = i + 1
	ogrenciNot = int(input(str(i) + ". ogrencinin notunu girin:"))
if i > 1:
	print("Ogrencilerin ortalamasi:" + str(toplam / (i - 1)))
else:
	print("Ortalamayi hesaplayacak kadar ogrenci yok")
```
