```
N = int(input())
bolunurMu = False
for i in range(2, N):
	if N % i == 0:
		bolunurMu = True
		break
if bolunurMu == False:
	print("Asal")
else:
	print("Asal Degil")
```

```
toplam = 0
N = int(input())
M = int(input())
i = 1
while i <= N:
	if i % M == 0:
		i = i + 1
		continue
	toplam += i
	i = i + 1
print(str(toplam))
```

```
toplam = 0
N = int(input())
M = int(input())
for i in range(1, N + 1):
	if i % M == 0:
		continue
	toplam += i
print(str(toplam))
```
